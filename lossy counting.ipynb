# Lossy Counting 

---

### Counting in data streams 

- many data mining tasks rely on counting: frequent patterns, similarity, correlation, probabilistic models, etc.
- Again, if the vocabulary is small, just keep a dictionary
- a smarter way is needed if vocabulary is large
- usually lossy, but with bounded errors 

- approximate answers are often sufficient (e.g., trend/pattern analysis)
- how much error is acceptable?
    - example from Prof. Jiawei Han: a router is interested in all flows whose frequency is at least 1/10 of $\sigma (\epsilon = 0.1%)$ error is comfortable 
- How to count the frequency of items in a data stream, with a limmited memory capacity and an error bound ($\epsilon N$)?

### Lossy counting in data streams 

- divide the data stream into buckets that can fit into memory 
- keep a summary of the counts in the stream
- count the frequency of items in each bucket 
- add item frequency of local buckets into the summary 
- discount them (so infrequent items will be gone)



- Divide the data stream into buckets of size $1/\epsilon$

### First bucket 



- count the frequency of items in bucket 1, add to summary (empty)



- drop the counts of every item by 1, so infrequent ones are gone 
- anti-smoothing? 

### Second bucket 



- count the frequency of items in bucket 2, add to summary 



- drop the counts of every item by 1: dictionary size is still small 

### Third bucket 



- count the frequency of items in bucket 3, add to summary 

### Summary after the third bucket 



- summary is compact (small dictionary, fit in memory)
- preserves frequent items (but false positives are possible)
- approximates the real counts 

### How good/bad is the approximation?

- after each bucket, every item is discounted at most by 1
- bucket size: $1/\epsilon$
- number of buckets: $N/(1\epsilon) = \epsilonN$
- so the error of count is at most $\epsilonN$
- this works well if $\epsilon$ is small 

### Summary of lossy counting 

- Strength: 
    - A simple idea, easy to implement 
    - No false negative 
    - Can be easily extended to handle frequent patterns 
- Weakness: 
    - many false positives
    - space bound is not good 
    - does not handle "locally" frequent items 

### Lossy counting in practice 

- process multiple buckets if memory allows 
- if counting k buckets at a time, reduce counts by k
    - remove more items
- use apriori: if a pattern is not frequent in summary, do not consider its super patterns 
- works well for concept drift, because items that become frequent more recently are more likely to pass the filter 

### Beyond counting 

- many data mining functionalities are built upon counting 
- frequent patterns, similarities, probabilistic models, regression/predictions, outliers, etc.
- more complex data mining tasks can be soled using similar idea: keeping a synopsis structure (summary) of the stream and updating it online
- e.g., training a machine learning model online using stochastic gradient descent 

### What you should know 

- using synopsis structure to summarize data stream
- how Bloom filter and lossy counting work 
- many data mining tasks can be facilitated through an online summary of the data stream
