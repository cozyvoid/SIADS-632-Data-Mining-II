### Smoothing 

- Assign a small probability to n-grams that have never appeared
- So we can calculate the probability of any out-sample sequence
- If we assign probabilities to unseen events, that means we have to distract those of observed events
- This process is called "smoothing"

### Add "One" Smoothing

- Laplace smoothing
- For every word w in the vocabulary, add the count of (k+1)-gram ending with w by 1 so every word has appeared after $w_1 ... w_k$ for at least once.



- Many other smoothing methods available in natural language processing

### Trick: do calculation with logarithm

- When sequence is long: avoid multiplying many probs
- Do it in log space: mathematically equivalent (why?)
- Avoid underflow and loss of precision
- Faster
