# Data Mining II - Intro.

### Recap of Data Mining I

- Real world data can be represented as itemsets, vectors, matrice, **sequences**, **time series**, networks, **streams**, etc.
- Data mining outputs (functionalities) include **patterns**, **similarity/distances**, classification, clustering, **predictions**, outliers, etc.
    - Patterns and similarities are often used as building blocks of complex data mining output and as features for downstream machine learning tasks.
- Algorithms for mining itemset data, vector/matrix data, and sequence data
