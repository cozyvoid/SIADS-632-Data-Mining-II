### Interpreting sequential patterns

- Why are bigrams always more frequent than trigrams?
- Why are some n-grams more frequent than others?
- What would the next item be in a sequence?



### "Language models"

- Evaluate the probability of any sequence of words
- Explain how text data (e.g., a sentence) is generated
- Predict which word sequence is more likely to appear next
- Any sequence is possible, but some has a smaller probability...

### A unigram language modeling

- A special (irregular) dice: N faces, each corresponding to a word in the vocabulary
    > S = V = {"cat", "dog", "is", "to", ...}

- Some faces have higher/lower probabilities than others
    > P("the") > P("cat") > P("jaguar"), ...

### Generate single-word sequences

- Select a word to write by tossing a dice 
- Toss the dice again to "generate" a new word
- Result:
    - is
    - cat 
    - to 
    - to 
    - dog 
    - ...
