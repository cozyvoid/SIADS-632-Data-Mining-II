### Data modeling

- A new data mining functionality
    - Input: data representation
    - Output: a “model” of the data
- Definition of a “model” can be quite different
- In data mining, we often refer to a statistical model
- A statistical model helps us understand the underlying properties/mechanism of the data, draw conclusions, and make decisions

### Outlier Detection

- An observation is considered as an outlier if:

    - It's very different from other observations (simiilarity)
    - It deviates much from our understanding about the data (modeling)
    - It's very different from the predicted value (prediction)

### What you should know

- What do sequences, time series, and data streams have in common and how are they different?
- Prediction/forecasting is a common task for these types of data.
- (Statistical) data modeling provides a solution for prediction.
- It also providess a solution for outlier detection.
