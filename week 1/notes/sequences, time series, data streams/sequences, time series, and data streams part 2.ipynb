### Prediction and forecasting

- Prediction: given the observed data, make an inference (guess) about an unobserved piece of information
    - a data object, a property, an event, information about the population/another sample, etc.
- Forecasting: when the target of prediction is in the future
- In this course, we use them interchangeably if no ambiguity
- A common data mining functionality of sequences, time series, and data steams
