### Applications of HMMs

- *Natural language processing*: part of speech training, entity/relation extraction, machine translation...
- *Bioinformatics*: gene detection, sequence alignment, structure prediction, motif discovery...
- Speech recognition, signal processing, time series analysis
- *Behavior analysis*: finance, sports, business, transportation...
- More powerful solutions available (deep learning, recurrent neural networks), but HMMs are still widely used as reliable baselines and building blocks for more complex techniques
