### Hidden Markov Model (HMM)

- Every time in a sequence is associated with a state
- Events/items are discrete, sequential
- States are finite, countable, but only partially observed

### Sequence with hidden states





### The generating process of HMM

- Observed: an **output** sequence



- Hidden (partially observed): a **state** sequence



- Every **state** only depends on the previous **state** (Markov)



- Once **state** is determined, the **output** does not depend on the previous observations
