1. Readability First — The Pythonic Way
Python emphasizes clarity over cleverness. Follow PEP 8 style guidelines:
Python

```    
# Good
total_price = price * quantity
```
```
# Bad
tp = p*q
```

✅ Use descriptive names, consistent indentation, and spaces around operators.

2. Use Python’s Built-in Functions
Instead of manual loops:
```
# Instead of:
total = 0
for num in numbers:
    total += num
```
```
# Do:
total = sum(numbers)
```

✅ Built-ins like sum(), max(), min(), sorted() are optimized and readable.

3. List / Dict / Set Comprehensions
Compact and expressive:

Python
```
# List comprehension
squares = [x**2 for x in range(10)]
```
```
# Dict comprehension
word_lengths = {word: len(word) for word in ["apple", "banana"]}
```
```
# Set comprehension
unique_lengths = {len(word) for word in ["apple", "banana", "apple"]}
```
✅ Avoids verbose loops, keeps logic in one place.

4. Multiple Assignment & Unpacking
Python
```
a, b = 5, 10
a, b = b, a  # swap without temp variable
```
```
# Unpack lists
first, *middle, last = [1, 2, 3, 4, 5]
```

✅ Improves clarity and reduces boilerplate.

5. Use enumerate() and zip()
Python
```
# enumerate: index + value
for i, val in enumerate(["a", "b", "c"], start=1):
    print(i, val)
```
```
# zip: iterate in parallel
names = ["Alice", "Bob"]
scores = [85, 92]
for name, score in zip(names, scores):
    print(f"{name}: {score}")
```

✅ Avoids manual index tracking.

6. Logical Reasoning Patterns

Short-circuit evaluation:

```
if user and user.is_active:
    print("Active user")
```
Any / All:
```
if all(x > 0 for x in numbers):
    print("All positive")
if any(x < 0 for x in numbers):
    print("At least one negative")
```
✅ Makes intent explicit.

7. Safe File Handling
```
with open("data.txt", "r") as f:
    content = f.read()
```
✅ with ensures files close automatically, even on errors.

8. Exception Handling
```
try:
    value = int(input("Enter a number: "))
except ValueError:
    print("Invalid number")
```
✅ Catch only specific exceptions to avoid masking bugs.

9. Debugging & Introspection
```
import pprint
pprint.pprint(vars(some_object))  # Pretty-print object attributes
```
✅ Helps understand unfamiliar code quickly.

10. Use pathlib for File Paths
```
from pathlib import Path
data_dir = Path("data")
for file in data_dir.glob("*.txt"):
    print(file.read_text())
```
✅ Cleaner than os.path and cross-platform.

Pro Tip for Reading Code
When reading unfamiliar Python code:

Identify imports — know what’s built-in vs. external.
Trace data flow — follow variables from creation to usage.
Look for patterns — comprehensions, decorators, context managers.
Run small parts in an interactive shell to see behavior.
