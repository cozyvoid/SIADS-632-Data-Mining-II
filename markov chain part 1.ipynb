### Generation process of bigrams

- Generating one word at a time!
- The first word in a sequence doesn't depend on anything
- The second word depends on the **first** word
- The thirs word depends on the **second** word
- ...
- The $n^th$ word depends on the **$(n-1)^th$** word



### The Chain rule still works

- How is this connected to the chain rule?



- Assumption: probability of writing one word down only depends on the previous word (if there is one)



### A Markov chain

- This is known as a "Markov" assumption
- Named after mathematician Andrey Markov
- A Markov chain: in a sequence of possible events, probability of each event depends onlhyh on the state of the previous event
