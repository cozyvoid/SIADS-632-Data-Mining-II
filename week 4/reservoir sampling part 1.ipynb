### Reservoir sampling 

- First K events will all be included into the sample 
- for the Nth event (N>K), include it into the sample with p = K/N 
- If so, randomly replace one event already in the sample 

### How it works 













### Why it works 

- The Nth event, $X_N$ (current event): $p = K/N$ 
- The Mth event, $X_M$ (an old event, $K \leq M < N$):
    - At time $t_M$: $p = K/M$ 
    - At time $t_{M + 1}$: there is a probability that it'll be replaced 
    - So the probability of remaining in the sample is: 





### What have we just found?

- At step $M$, if the probability of **any** event being in the sample is $K/M$, then at step $M+1$, the probability that the same event is in sample is $K/(M+1)$
- The probability of older events staying in sample decreasess over time and matches that of the new events!

### The proof 

- Let $P_{i,N}$ be the probability that even $X_i$ is in sample at time $t_N$
- $P_{i,i} = min(K/i, 1)$
- Proposition: for all $N \geq K, P_{i,N} = K/N for all i \leq N$
- Let's call it statement $A(N)$

### The proof (mathematical induction)

- the statement is true for $N=K$, or $A(K)$ is true 
- (When $N = K$, **for all** $i \geq K$ we have $P_{i,K} = 1 = K/K$)
- suppose $A(M)$ is true for some $M \geq K$, is $A(M + 1)$ also true? 
    - for $i = M + 1$, we have $P_{i,M+1} = K/(M+1)$
    - for $i < M + 1$, we have :
    
    

    - $A(M+1)$ is true
