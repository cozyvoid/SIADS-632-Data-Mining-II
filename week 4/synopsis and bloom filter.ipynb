# Synopsis and Bloom Filter 

---

### Challenges of mining stream data 

- One example question that summarizes all the challenges: 
    - 500 million Tweets being generated per day
    - You have the capacity to keep 1000 tweets (at a given time)
    - How do you do data mining? 

### Sampling solves some of them 

- instead of dealing with a huge stream, you now deal with a sample of limited size 
- at any given time, the stream essentially becomes a set, sequence, or time series 

### Limitation of sampling 

- K (number of events in sample) is usually much much smaller than N (number of events in the entire data stream)
- it's hard for a small sample to preserve more complex statistical properties 
- individual events can be large and complex
- keeping statistics can be more efficient than keeping events 

### Alternative to sampling events

- summarize the data stream with statistics
- keep statistics such as appearance, counts, mean, variation, and higher-order moments 
- updating these statistics over time 
- beyond simple statistics: sketches, models, synopses, etc.

#### Synopsis of data streams 

- also known as "summary"
- usually more efficient than keeping events
- can be more informative than a small sample of events 
- usually rely on randomized algorithms
- usually infeasible to give accurate answers, but can guarantee some error bounds 

#### Appearance 

- check whether an item has ever appeared in the stream
    - e.g., has anyone used the word "serendipity" in the past year?
    - has James Bond ever conneced to the server?
- if the vocabulary (number of unique items) is small, just keep a dictionary
- if the vocabulary is too large, use Bloom filter 

### Bloom filter 

- based on multiple hash functions
- keep an array of N bits
- a hash functions maps an item into K slots - a fingerprint 
- to query an item, check whether its fingerprint is present

### bloom filter in practice

- possible false positive, but no false negatives 
- output: "possible present" or "definitely not present"
- provable upper bound for false positives
- many extensions to lower the upper bound
- does not work for counts
