### Reservoir Sampling (once more)

- New item has probability K/N to be sampled 
- If sampled, replace an old item in the sample at random
- All items: equal probability of being sampled (K/N)

### Handling concept drift 

- Recent events should carry more weights because:
    - recent observations may be more relevant to decision making 
    - data generation processs may have changed
    - context may have changed 
- When sampling, assign a higher probability to more recent events to account for concept drift.

### Weighted reservoir sampling 

- Assigning a higher probability to a more recent event 
- Older events have smaller probability, but they can still be sampled (instead of only keeping most recent events)

- Design the process so that the probability of event $X_i$ to be included in the sample is proportional to a funciton $f(i,N)$

### Reservoir sampling in practice 

- often do not put hard constraints on the sample size, so an old event may or may not be kicked out (especially to facilitate exponential decay)
- or multiple events may be kicked out
- for certain policies, the bounds of the sample size is guaranteed with proof
