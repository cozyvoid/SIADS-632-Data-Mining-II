# Data streams 

- A list of ordered or timestamped **data objects, or "events"**



### What they have in common 

- observations are ordered
- next observations depends on previous observations 
- prediction/forecasting is a common task 
- modeling this sequential dependency is the key 
    - Markov models, ngram models
    - AR, MA, ARMA, ARIMA models 
    - Edit distance and Dynamic Time Warping
    - Moving averages, reservoir sampling 
- Outlier detection: observations that deviate "too much" from the "model"

### Differences

- sequence data are categorical: probabilistic models 
- time series data are numerical: regression models 
- data stream: online updates 

- order is discrete: ngrams, sequential patterns
- time is continuous: trends, seasons, cycles 
- transform between continuous and discrete spaces: sampling, regression, Fourier transforms, etc.

### Beyond sequential representations

- the basic principle of dealing with sequences, time series, and data streams is to handle sequential dependency
- this principle can be found in many more complex data mining models and tasks, e.g., recurrent neural networks, natural language processing 
- other data representations have more complex dependency structures: e.g., spatiotemporal data, networks, images, etc.
