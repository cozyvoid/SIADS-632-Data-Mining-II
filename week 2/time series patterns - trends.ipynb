### Patterns in time series 

- **Trends** *(T)*: long-term movements 
- **Cyclic movements** *(C)*: long-term oscillations about a trend line or curve
- **Seasonal movements** *(S)*: periodical patterns
- **Irregular movements** *(I)*: anomalies, outliers, randomness...

### A simple view of time series patterns 

- Decomposition of a time series into these four basic movements 

    1. **Additive Model**: $X = T + C + S + I$
    2. **Multiplicative Model**: $X = T \times C \times S \times I$ 

### Trends 
 
- Also known as "trend curve"
- General direction in which a time series is moving over a long interval of time 



#### Discover trends 

- The **freehand method**
- **Regression-based methods**
- **Smoothing-based methods** 

##### Free-hand methods
- Fit the curve by looking at the graph
- Costly and barely reliable for large-scale data mining 

#### Regression-based methods
- Also referred to as **least-squares** methods
- Find the curve by minimizing the sum of the squares of the deviation of points on the curve from the observed points
- The shape of curve is sensitive to the choice of regression 

- Ex: fit curves with least square
    - Data: monthly number of international airline passengers from year 1949-1960 (Box & Jenkins airline data)

##### Smoothing-based methods 
- also known as "noise reduction"
- Binning
- Moving averages
- Autoregressive methods 

##### Binning
- Divide the time stamps into bins (e.g., minutes -> hour, days -> year), and summarize the data into each bin





##### Problem of Binning

- Losing data points: number of binds are much smaller than number of measurements
- Boundaries of bins are arbitrary
- Can hide many useful patterns
