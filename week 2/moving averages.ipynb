# Moving Averages

---

### Moving Average

- Basic idea: smooth the data using **previous** observations
- (Sometimes people use observations before and after)
- Also known as "sliding window" average/smoothing
- Overlapping bins
- Think about connection to ngrams?
- Could eliminate cyclic, seasonal and irregular movements 

### Examples of moving average 





###  Moving Average in practice

- In time series forecasting: use **previous** observations only, don't look into the future
- Preserve the lenth of a series, but lose data at the beginning
- Sensitive to the selection of window size 
- Sensitive to outliers (can be reduced by weighted average)

### Cumulative moving average  

- Weighted moving average is still sensitive to the window size 
- Getting rid of the window size:
    - Summarizing **all observations** up to now 



### Computing cumulative moving average

#### Cumulative moving average 

- Window size is no longer arbitrary
- More robust to "recent" outliers
- But if there is one, it's going to have an effect forever
- Same effor for a recent observation vs. an ancient observation 



### Weighted cumulative moving average 

- If we weigh observations properly, window size is not an issue and can control the effect of historical observations
- The current observation should be respected
- More recent observations are more important
