### Time series similarity / distance 

- A basic data mining question: given two time series, how similar are they? 



### Time series similarity



### Similarity search 

- Normal database queries find exact matches
- Similarity search finds data series that differ only slightly from the given query series 
- Two use scenarios: 
    1. **Whole series matching**: find a series similar to the entire query series
    2. **Subsequence matching**: find a series if the query matches part of the series 

### Euclidean distance 

- Treating each time series as a *vector* 
- Assumption: two series have the same number of dimensions
- Timestamps (or positions) are naturally matched 



### Other distance metrics applicable 

- Manhattan distance 


- Cosine similarity
- Pearson's coefficient 
- ... 

### Problems of Euclidean distance 

- Sensitive to scale
    - Trick: always normalize the time series first (preserves the patterns)
    - Or use cosine similarity or Pearson's coefficient
- Two time series are not always aligned!
