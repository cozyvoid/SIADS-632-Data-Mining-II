### Time series alignment 

- Think about two runners, one running faster than the other
- To align the trip, the faster needs to "wait" for the slower
- Sometimes the other runner may be faster and needs to "wait" in return

### Principles of time series alignment 

- Every point in one time series mush be matched with one or more points in the other time series 
- The first point ($x_1$) must be matched with the first point ($y_1$), and maybe a few more 
- The last point ($x_n$) must be matched with the last point ($y_m$), and maybe a few more
- No "breaking orders" - e.g., if $x_i$ is matched with $y_j$, then $x_{i+1}$ cannot match any points before $y_j$, and vice versa 

### Basic idea of time series alignment 



### Optimized time series alignment 

- Whenever two points are matched, there is a cost $d(x_i, y_j)$
- Find the alignment that gives the smallest sum of costs
- The sum of costs provides the distance between the two series 

### Dynamic Time Warping (DTW)

- The best alignment up to the current time index depends on the alignment up to the previous time index
- Choose from three actions: X waits for Y; T waits for X; or they both proceed
    - X waits for Y: extend X by repeating the previous observation of X 
    - Y waits for X: extend Y by repeating the previous observation of Y 
    - Both proceed: match new observations of X and Y 



### Dynamic Programming 

- The definition of DTW (i, j) is recursive, which depends on DTW(i-1, j), DTW(i, j-1), and DTW(i-1, j-1)
- A straightforward way is to implement it in recursion
- Recursions are costly...
- Always use dynamic programming!

- Starting from the most simple case: DTW(1, 1)
