### Symbolic representation

- Symbolic Aggregate Approximation (SAX)
- Developed by Prof. Eamonn Keogh
- Lower bounding of Edit distance and DTW
- Fast: transform takes linear time
- http://www.cs.ucr.edu/~eamonn/SAX.htm



### Summary of time series similarities

- Analogy to vector data: Euclidean distance, Pearson's correlation, cosine, etc.
    - They are often applied to the results of DFT or DWT 
- Analogy to sequence data: edit distance, DTW, symbolic representations 
- They can all be applied after smoothing
