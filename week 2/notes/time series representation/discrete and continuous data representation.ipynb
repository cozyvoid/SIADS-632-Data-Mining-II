# Discrete vs Continuous Data Representations 

---

### Why time series?

- Discrete vs. continous data attributes
- Discrete data: values are finite or countable
- Continous data: values are infinite and not countable 

### Discrete vs. Continuous attribute

- Discrete attributes:
    - binary values (0/1, T/F, etc.)
    - categories (e.g., names, items, nodes, edges, etc.)
    - ordinal dta (e.g., rating 1-5, scales, etc.)
    - countable numbers (e.g., ages, year, integers, etc.)
- Continuous attributes:
    - real numbers (e.g., temperature, RGB colors, probabilities, etc.)
- That's why we use vectors instead of itemsets and sequences, no?
