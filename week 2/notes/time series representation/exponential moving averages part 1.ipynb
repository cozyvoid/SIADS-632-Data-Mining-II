### Exponential Moving Average (EMA)

- Weighted moving average, where weight decays exponentially 
- Interpolating observed value and a smoothed previous value 





### Example of EMA's



### How to interpret $\alpha$?

- $\alpha$ is often known as the "decay factor"
- By every unit of time, an old value declines by $\alpha (0 \leq \alpha \leq 1) $
- $\alpha = 1$: no smoothing
- $\alpha = 0$: no decay, uniform value ($x_1$)
- Different from "decay rate": $ -ln(1 - \alpha)$
- In practice, this is chosen empirically (try multiple values)

### Another way to look at EMA

- Every observation has an impact to the infinite future
    - e.g., profit and loss
- But the impact *decays* over time (exponentially)
- EMA is the *weighted average* of all historical values, by their remaining impact 
- Analogy: radiation, survival
- Half life: time required for the impact to reduce in half
