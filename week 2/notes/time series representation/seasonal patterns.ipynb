# Seasonal Patterns 

---

### Seasonal patterns 

- Almost indentical patterns that a time series appears to follow during a predefined period
- Naturally aligned to hours-of-day, day-of-week, months-of-year, etc.



### Examples of seasonal patterns 





### Simplest seasonal patterns: waves





### Different types of waves



### Mixture of waves 

- Waves of different frequency and amplituide can be mixed together to obtain more complex seasonal patterns 



### Reverse engineering

- Can a complex seasonal pattern be decomposed to a mixture of waves? If so, the frequences and the amplitudes of the waves can describe what are being repeated over time. 
- Yes we can. It is guaranteed by the **Fourier theorem**.

### The Fourier Theorem

- Any **periodical** function $f(x)$ can be written into the sum of a series of sin (and cosine) waves with various frequencies and amplitudes.
- $f(x)$ has to be "reasonably" continuous 

### Fourier transform 

- Think of $f(x)$ as the continuouss representation of a time series
- Then the Fourier Theorem guarantees that we can decompose any time series with seasonal patterns into a mixture of simple periodical patterns (waves)
- This can be done through **Fourier transform**.











### Time domain and frequency domain 

- Time domain representation: how signals change over time
- Frequency domain representation: how much the changes follow certain frequencies 



### Transform between time and frequency domains 

- On either time or frequency domains, a representation can be either continuous or discrete 







### Use DFT/FFT for time series 

- Input: a time series (represented in time domain)
- Output: a Fourier series (a vector)
- Dimensions are ordered, corresponding to different frequencies
- High amplitude on one dimension indicates there is a seasonal pattern appearing at the corresponding frequency 
- The vectors can be used to extract more complex features (recall mining matrix data)

### Seasonal index 

- Once seasonal patterns are discovered, we can rearrange the time series to reflect the seasonality
- e.g., set of numbers showing the relative values of teh sales during the months of the year
- If the sales during October, November, and December are 80%, 120%, and 140% of the average monthly sales for the whole year, respectively, then 80, 120, and 140 are seasonal index numbers for these months
