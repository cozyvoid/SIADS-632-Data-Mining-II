### Related concepts (in radiative decay)

- **Quanity** ($N_t$): how many particles are there at time t
- **Decay factor** ($\alpha$): proportion of particles decayed by next day 
- **Decay rate** ($\lambda$): $-ln(1 - \alpha)$
- **Half life** $(t_{1/2})$: time at which *half of the particles* decay 
- **Mean lifetime** $(\tau)$: average life time of an individual particle

### Autoregressive methods 

- More complex than moving average
- Can be interpreted as a moving average method where the weights $(w_i)$ of previous observations are not predefined, but they are estimated through a regression
- Can also be interpreted as using the past k observations to predict the next observation
