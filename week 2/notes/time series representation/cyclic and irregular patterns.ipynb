### Cyclic patterns

- May or may nore be periodic or repeating in a single time series

### Extract Cyclic Patterns 

- Extracting cyclic patterns can be harder than seasonal patterns, as the patterns aren't necessarily periodical 
- If (approximate) periodicity of cycles occurs, they can be constructed similarly to seasonal patterns 
- If not, needs to use other methods such as wavelet analysis 

### Wavelet Analysis 

- Problem with Fourier transforms: gives information about which frequencies appear, but no information about when they appear (location in time series)
- Why is this important for cyclic patterns?
- Wavelet transform: preserves information of both the (time) location and the frequency of a signal 

### A demo of wavelet transform 

- Continuous wavelet transform
- Discrete wavelet transform (widely used for time series)



### Wavelet transforms for time series 

- Input: time series (represented in time domain)
- Output: a frequency * time matrix (or a heatmap)
    - columns correspond to timestamps, ordered
    - rows correspond to frequencies, ordered 
- Can be used to extract more complex patterns (recall mining matrix data)



### Irregular Movements

- Extracted by adjusting the data for trend, seasonal and cyclic variations
- Usually extracted from the residual of a time series model 
- Additive model: $X = T + C + S + I$
- Multiplicative model: $X = T \times C \times S \times I$
- Anomalies, outliers
- Random movements 
- Noise
