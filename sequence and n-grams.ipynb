# Sequence and N-grams

---

### Describe the properties of n-grams

- **Support**: how many records/sequences contain an n-gram (normalized by the # of records)
- **Count**: how many times an n-gram appears in a database (it could appear multiple times in one sequence)
- **Frequency**: count divided by the total number of n-grams, sometimes used equivalently as count 

### Interpreting the patterns

- Why are bigrams more frequent than trigrams?
- Why are some n-grams more frequent than others?
- What would the next item be in a sequence?
